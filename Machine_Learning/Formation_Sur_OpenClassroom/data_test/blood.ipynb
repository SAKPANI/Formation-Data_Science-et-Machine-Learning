- Gestion des outlier

```python
Q1 = projet["budget_demande_fcfa"].quantile(0.25)
Q3 = projet["budget_demande_fcfa"].quantile(0.75)

IQR = Q3 - Q1


born_inf = Q1 - 1.5*IQR
born_sup = Q3 + 1.5*IQR

outliers = projet[(projet["budget_demande_fcfa"]<born_inf) | (projet["budget_demande_fcfa"] > born_sup)]
```
