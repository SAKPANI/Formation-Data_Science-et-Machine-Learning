<p style="text-align:justify;">
Dans le domaine de la <strong>Data Science</strong>, les données proviennent de nombreuses sources (fichiers CSV, fichiers Excel, bases de données, API, etc.) et se présentent sous différentes formes. Afin de les exploiter efficacement, il est indispensable de disposer d'un outil capable de les importer, de les organiser, de les nettoyer, de les transformer et de les analyser.
</p>

<p style="text-align:justify;">
C'est précisément le rôle de la <strong>bibliothèque Pandas</strong>, l'une des bibliothèques les plus utilisées en Python pour la manipulation et l'analyse de données. Elle fournit des structures de données puissantes et intuitives qui permettent de travailler efficacement sur des jeux de données de toute taille.
</p>

<p style="text-align:justify; color:#c0392b;">
Le <strong>DataFrame</strong> est la structure de données principale de Pandas. Il permet de représenter les données sous la forme d'un tableau à deux dimensions, composé de lignes et de colonnes dont chacune possède un nom explicite. À l'image d'un tableau Excel ou d'une table d'une base de données, chaque colonne représente une variable et chaque ligne correspond à une observation (ou individu).
</p>

<p style="text-align:justify;">
Chaque colonne d'un <strong>DataFrame</strong> possède un type de données unique (<em>entier</em>, <em>réel</em>, <em>chaîne de caractères</em>, <em>date</em>, etc.), tandis que différentes colonnes peuvent contenir des types différents. Cette organisation rend les données faciles à manipuler et permet d'effectuer de nombreuses opérations telles que le filtrage, le tri, les calculs statistiques, la fusion de tableaux, le traitement des valeurs manquantes et la visualisation des données.
</p>

Au cours de cette formation, nous découvrirons progressivement les principales fonctionnalités de Pandas afin d'acquérir les compétences nécessaires pour manipuler des données réelles et réaliser des analyses efficaces dans le cadre de projets de Data Science.

<h3 style = "color:red">3. Filtrer avec des conditions </h3>

```python
Etudiant = {
    "Nom" : ["SAKPANI", "PAGUI", "HRO", "DPYRO", "gedon"],
    "Filiere" :["IA", "GC", "GE", "LT", "GM"],
    "MATHS" :[18,14,10,2,13],
    "PHY" :[14,13,4,0,9]
}
df = pd.DataFrame(Etudiant)
df
```

```python
df[df["MATHS"]>10]  ##Selectionne les ligne qui respecte la conditon MATHS>10
```

```python
df
```

```python
df[df["PHY"] < 10]   ###AFFICHE LES DONNES SELON LA CONDITION PHY >10
```

```python
df[(df["MATHS"] > 10) & (df["PHY"] >10)]   ## Affiche lese etudiant qui on 10 de plus en physique et msths
```

```python
df["moyenne"] = (df["MATHS"] + df["PHY"])/2  #ajout de la moyenne dans le dataSet
```

```python
df
```

```python
df[df["moyenne"]<10]  ##Affiche les etudiant ayant otenu des moyenne inf a 10
```
