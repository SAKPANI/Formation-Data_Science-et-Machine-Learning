## Modifiez les éléments extérieurs

Dans un premier temps, concentrons-nous sur les aspects “extérieurs” d’un graphique. Si nous reprenons le graphique du chiffre d’affaires total réalisé par agence :

Vous noterez qu’il manque quelques grilles de lecture :

```python
df = data.groupby("ville")["remboursement"].sum()
df = df.reset_index()
df 
```

```python
plt.figure(figsize = (18 , 8))  # Creation de la fenetre de nos figure

plt.bar(height  =df["remboursement"] ,x = df["ville"])

plt.title("Chiffre d'affaire realisee par agence", fontsize = 19)

plt.xlabel("Agences", fontsize = 19)

plt.grid(axis = "y") # Grille uniqument sur l'axe y 

plt.ylim(0, 120000) # Limite des graduations (pour l'axe dews ordonnes 

#plt.xlim(0, 13) # limite de graduation pour l'axe des abscisses 
# plt.yticks(debut , fin , nombreElemnt) indique les graduations

plt.ylabel("Chiffres d'affadires", fontsize =18)

ca = df["remboursement"].tolist()
for i in range(len(ca)):
    plt.text(i-0.2, ca[i]+2000,round(ca[i], 2))  # + 2000 crer une marge survant l'ae y , et -0,2 permet de reculere suvsnt l'abscisse x
plt.show()
```

```python

CA = pd.read_csv('https://raw.githubusercontent.com/OpenClassrooms-Student-Center/fr-4452741-decouvrez-les-librairies-python-pour-la-data-science/main/data/CA.csv')
CA['date'] = pd.to_datetime(CA['date'])
CA.head()
```

```python
type_donne = CA.columns.tolist()
type_donne =  type_donne[1:]
type_donne 
```

```python
plt.figure(figsize =(6, 5))
for type in type_donne :
        plt.plot(CA["date"], CA[type], label = type)
plt.xlabel("dates")
plt.ylabel("Benefices")
plt.title("Benefice mensuele de l'a annes 2021 par type de prets", fontsize = 14)
plt.grid()
plt.yticks(fontsize=11)  ## Taille de la graduiation  11
plt.legend()
```
