```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
# import du fichier client_assurance netoyee
df= pd.read_csv("fichier_netoyee.csv")
df.head()
```

<h1 align="center" style="color:blue;">
 MATPLOTLIB
</h1>

```python
plt.plot(df["age"], df["prime"], label = "prime = f(age)")
plt.title("Prime en fonction de l'age")
plt.grid()
plt.xlabel("age")
plt.ylabel("prime")
plt.legend()
plt.show()
```

<h2 style ='color : blue'>A. Nuages de point (Relation entre deux variables) </h2>

<p>Pour analyser la relation entre deux variable on utilise les nuages de point.<br>La fonction qui permet de repenter les nuages de point avec la bibliotheque matplotlib est la fonction <span style = "color  : blue ">plt.scatter(var1, var2)</span></p>

```python
plt.scatter(df["age"], df["prime"], color ="blue")
plt.xlabel("age")
plt.ylabel("prime")
plt.title("Relation entre l'age et prime")
##plt.show()
```

```python
## d'APRES L'HISTOHRAMME PLUS L'AGE S'AUGMENTE , IL Y;A DIMUNITION DE PRIME
# - D'apres les nuages de points les jeunes obtiennes plus de primes que les vie
```

```python
##Corelation entre age et prime
```

```python
Coorelation  = df["age"].corr(df["prime"])
print("Correlation entre age et prime : ",float(Coorelation))
```

### Petite remarque sur les correlation :

#### - entre 0 et + - 0.2 ==> Correlation quasi inexistante
#### - Entre +-0.2 et +-0.4 ==> Correlation faible a modérée
#### - Entre +- 0.4 et +- 0.7 ==> Correlation moderee a forte 
#### - Au dela de +-0.7 ==> Correlation forte

```python
plt.hist(df["age"], bins =20, label = "age")  ##Decoupage en 20 histogramme (Par defaut python desssine 10)
plt.title("Distribution de l'ages des clients de l'assurance")
plt.legend()
plt.show()   ##La  distribution des ages 
```

<h1>Lire et interpréter les graphiques</h1>
<p class="sous-titre">Matplotlib / Seaborn — dataset clients_assurance.csv</p>

<h2>1. Histogramme — distribution d'une variable</h2>
<pre><code>plt.hist(df['age'], bins=20)
plt.show()</code></pre>

<div class="bloc lire">
<span class="label">Comment le lire :</span>
Imagine 305 fiches clients rangées dans 20 boîtes selon leur âge. Chaque barre = une boîte, et sa hauteur = le nombre de fiches dedans. Une barre haute = beaucoup de clients à cet âge, une barre basse = peu de clients. Il ne montre <b>qu'une seule variable</b> à la fois.
</div>
<div class="bloc utilite">
<span class="label">Utilité :</span>
Voir comment une population est répartie (beaucoup de jeunes ? de seniors ? une répartition équilibrée ?). Le premier réflexe sur une variable numérique.
</div>
<div class="bloc piege">
Piège : un histogramme ne peut pas montrer de relation entre deux variables (âge et prime par exemple). S'il y a deux axes avec des points, ce n'est pas un histogramme mais un nuage de points.
</div>

<h2>2. Nuage de points (scatter) — relation entre deux variables</h2>
<pre><code>plt.scatter(df['age'], df['prime'])
plt.show()

# Version Seaborn avec une 3e variable en couleur
sns.scatterplot(data=df, x='age', y='prime', hue='type_contrat')</code></pre>

<div class="bloc lire">
<span class="label">Comment le lire :</span>
Chaque point = un client. Position horizontale = âge, position verticale = prime. Si les points descendent de gauche à droite, la prime baisse quand l'âge augmente. S'ils montent, c'est l'inverse. Un nuage sans direction = pas de lien visible.
<br><br>
Avec <code>hue='type_contrat'</code>, chaque point est en plus coloré selon son type de contrat — on voit alors trois informations à la fois sur un seul graphique.
</div>
<div class="bloc utilite">
<span class="label">Utilité :</span>
Détecter une relation entre deux variables numériques, et repérer les points isolés loin du nuage principal (outliers).
</div>
<div class="bloc piege">
Piège : une tendance vue à l'œil reste une impression, pas une preuve. Pour la confirmer, on calcule la corrélation chiffrée :
<pre><code>df['age'].corr(df['prime'])</code></pre>
</div>

<table>
<tr><th>Valeur (absolue)</th><th>Interprétation</th></tr>
<tr><td>0 à 0.2</td><td>corrélation quasi inexistante</td></tr>
<tr><td>0.2 à 0.4</td><td>corrélation faible à modérée</td></tr>
<tr><td>0.4 à 0.7</td><td>corrélation modérée à forte</td></tr>
<tr><td>0.7 à 1</td><td>corrélation forte</td></tr>
</table>
<p>Le signe (+ ou -) donne la direction : négatif = les variables évoluent en sens opposé, positif = dans le même sens.</p>

<h2>3. Boxplot (boîte à moustaches) — comparer une variable par catégorie</h2>
<pre><code>sns.boxplot(data=df, x='type_contrat', y='prime')
plt.show()</code></pre>

<div class="bloc lire">
<span class="label">Comment le lire :</span>
Les fiches sont séparées en tas selon la catégorie (ici le type de contrat). Chaque tas devient une boîte qui résume : la ligne au milieu = la médiane des primes ; la hauteur de la boîte = où se trouvent 50% des clients ; les points isolés au-dessus/en dessous = des outliers.
</div>
<div class="bloc utilite">
<span class="label">Utilité :</span>
Comparer une variable numérique entre plusieurs groupes en un coup d'œil, et repérer les valeurs extrêmes par groupe.
</div>
<div class="bloc piege">
La boîte la plus haute = la catégorie avec la prime médiane la plus élevée. Une boîte large = primes très variées dans ce groupe ; une boîte fine = primes proches les unes des autres.
</div>

<h2>4. Barplot (diagramme en barres) — comparer des effectifs ou des moyennes</h2>
<pre><code># Effectifs : combien de clients par categorie
df['type_contrat'].value_counts().plot(kind='bar')

# Moyenne par categorie (Seaborn)
sns.barplot(data=df, x='type_contrat', y='prime')</code></pre>

<div class="bloc lire">
<span class="label">Comment le lire :</span>
Chaque barre = une catégorie, sa hauteur = un comptage (nombre de clients) ou une moyenne, selon la fonction utilisée. Avec <code>sns.barplot</code>, un petit trait vertical en haut de la barre indique la marge d'incertitude sur la moyenne calculée.
</div>
<div class="bloc utilite">
<span class="label">Utilité :</span>
Répondre vite à des questions simples : quelle catégorie est la plus fréquente ? quelle catégorie a la moyenne la plus haute ?
</div>
<div class="bloc piege">
Le barplot ne montre qu'un seul chiffre résumé par catégorie, contrairement au boxplot qui montre toute la distribution.
</div>

<h2>5. Heatmap de corrélation — vue d'ensemble des relations</h2>
<pre><code>sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='Greys')
plt.show()</code></pre>

<div class="bloc lire">
<span class="label">Comment le lire :</span>
Un tableau où chaque case croise deux variables numériques. Le chiffre dans la case = leur corrélation (entre -1 et 1, même lecture qu'au point 2). Plus la case est foncée, plus la relation est forte. La diagonale vaut toujours 1 (une variable comparée à elle-même) — elle ne s'interprète pas.
</div>
<div class="bloc utilite">
<span class="label">Utilité :</span>
Voir toutes les relations entre variables numériques en un seul graphique, plutôt que de faire un scatter pour chaque paire une par une.
</div>
<div class="bloc piege">
Cherche les cases (hors diagonale) les plus proches de 1 ou -1 : ce sont les relations les plus fortes du dataset, celles à creuser en priorité.
</div>
