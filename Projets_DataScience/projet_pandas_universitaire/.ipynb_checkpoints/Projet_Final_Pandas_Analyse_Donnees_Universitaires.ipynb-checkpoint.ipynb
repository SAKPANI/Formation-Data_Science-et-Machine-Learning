#### 8 - ) Calcule de  la moyenne générale de chaque étudiant.

```python
Data_notes.head()
```

La formule de calcul de moyenne General : **MoyGeneral = somme(note_matiere x Credits)/somme(Credits)**

```python
SommeCredits = Data_notes["credits"].sum()
SommeCredits
```

Le La somme total des credit du semestre 1 et 2 : ***Somme_Credits = 28800 credits***

```python

Data_notes["Points"] = Data_notes["note"]*Data_notes["credits"] #calcul de points 
moy = Data_notes.groupby("student_id").apply(
    lambda g: g["Points"].sum()/g["credits"].sum()   #Creation de la fonction de calcule de moyenne et application au data_frame Data_notes
)

Data = moy.reset_index([0])  # Traduction en dataframe classique 
```

<h4 align = "center" style = "color : blue">TABLEAU CONTENANT LA MOYENNE GENERAL DE CHAQUES ETUDIANT</h4>

```python
moyGeneral = Data.rename({0:"moyenne_Generale"}, axis = 1)#J'ai renome la colonne qui contenait nos moyenne car par defeau le dataframe indiquai
moyGeneral=round(moyGeneral.loc[:, ["student_id","moyenne_Generale"]], 2) #arrondi a 2 valeur apres la virgule

moyGeneral.head() # j'AI SEULEMENT AFFICHER LES 5 PREMIER ETUDIANT DU DATAFRAME
```

<h4 align = "center" style = "color : blue">CLASSEMENT DES ETUDIANTS SELON LEURS MOYENNE GENERAL (LES 10 PREMIERS) </h4>

```python
Classement_general = moyGeneral.sort_values("moyenne_Generale", ascending = False)
Classement_general.head(10) # Les dix premiers meilleur moyenne generale
```

#### 10- et 11) Déterminer le nombre d'étudiants ayant une moyenne générale >= 10. et ceux <10

```python
nombre_moy_sup_10 = Classement_general.loc[Classement_general["moyenne_Generale"] >= 10, "moyenne_Generale"].shape[0] #Avec la methode des crochets 
nombre_moy_inf_10 = Classement_general.query("moyenne_Generale < 10").shape[0]                                             # Avec la  methode query

print(f"Nombre de moyenne >=10 : {nombre_moy_sup_10}")
print(f"Nombre de moyenne < 10 : {nombre_moy_inf_10}")
```

- Le nombre d'etudiant ayant une **moyenne générale >=10 sont : 598 Etudiants**
- Le nombre d'etudiant ayant une  **moyenne générale <10 sont  : 202 Etudiants**

## Partie 8 — Analyse avancée

Créer un indicateur de performance pour chaque étudiant.

Exemple de catégories :

- Très faible
- Faible
- Moyen
- Bon
- Excellent

Créer également un indicateur de risque :

- Risque élevé
- Risque moyen
- Risque faible

Les règles doivent être définies et justifiées.

```python
absent.head() # J'ai afficher absent dans cette partie car si je dois creer un indicaateur de risque j'aurai besoin desn taux de 
#precece de retard de se fait on doit fusionner ces donne avec lcelui de la moyenne general
```

### Creation de l'indicateur de performance

```python
moyGeneral.head()
```

```python
def categorie_moyenne(moyenne):
    if moyenne <8:
        return "Tres faible"
    elif moyenne >8 and  moyenne <10:
        return "Faible"
    elif moyenne<=14:
        return "moyen"
    elif  moyenne <16:
        return "Bon"
    elif moyenne>=16 and  moyenne<=20:
        return "Excellent"     
moyGeneral["Categorie"] = moyGeneral["moyenne_Generale"].apply(categorie_moyenne)

moyGeneral.head()
```

### Creation de l'indicateur de risque

```python
df_risque = pd.merge(absent, moyGeneral, on ="student_id", how = "inner")
df_risque.head()
```

#### Regles :
- Risque élevé : moyenne < 8 ET taux d'absence > 30%, retard <=2
- Risque moyen : moyenne entre 8-12 ET/OU absences entre 15-30%, retard entre 2 et 5
- Risque faible : moyenne ≥ 12 ET absences < 15%, retard >5

```python
def risque(moyenne, taux_absence, retard):
    # Points pour la moyenne
    if moyenne < 8:
        points_moyenne = 2
    elif moyenne <= 12:
        points_moyenne = 1
    else:
        points_moyenne = 0
    
    # Points pour le taux d'absence
    if taux_absence > 30:
        points_absence = 2
    elif taux_absence >= 15:
        points_absence = 1
    else:
        points_absence = 0
    
    # Points pour le retard
    if retard > 5:
        points_retard = 2
    elif retard > 2:
        points_retard = 1
    else:
        points_retard = 0
    
    # Score total
    score = points_moyenne + points_absence + points_retard
    
    # Classification finale
    if score <= 1:
        return "faible"
    elif score <= 3:
        return "moyen"
    else:
        return "élevé"

df_risque["risque"] = df_risque.apply(
    lambda applique_ligne : risque(applique_ligne["moyenne_Generale"],applique_ligne["absences"],applique_ligne["retards"]),axis = 1)
df_risque.head()
```
