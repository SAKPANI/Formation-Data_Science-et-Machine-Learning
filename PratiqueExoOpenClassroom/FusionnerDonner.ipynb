<h2 style = "color  : blue">La fonction ou  méthode ** merge() ** </h2>

merge() permet de fusionner des donnees d'un DATAFRAME en utilisant une clé ou une colonne de corresponance
### *Syntaxe :*
<p>- A.merge(B) , merge est considerer ici comme une methode<br> - pd.merge(A,B) , fonction pandas</p>

```python
clients = pd.DataFrame({"id_client": [1,2,3], "nom": ["Alice","Bob","Chloe"]})
clients # 
```

```python
# # merge() : jointure sur unecolonne commune

# 
# pd.merge(clients, contrats, on=’id_client’, how=’outer’) # garde tout, meme sans
# correspondance
```

```python
contrats = pd.DataFrame({"id_client":[1,2,4], "prime": [200,350,150]})
contrats
```

```python
Dataset = pd.merge(clients, contrats)
Dataset 
```

```python
# # En utlisant la fonction merge() , on remarque que le nouveau dataset affiche que deux ligne :
#    - En fait la fonction merge() fusionne les donnes en utilisant une correspondance du dataframe clients
```

```python
#Autre methode 
```

```python
 pd.merge(clients, contrats, on="id_client", how="inner") # seulement les orrespondances
```

```python
pd.merge(clients, contrats, on="id_client", how="left") # garde tout "clients
```

<h2 style = "color  : blue">La fonction ** concat() ** </h2>

```python
df_2023 = pd.DataFrame({"nom": ["Alice"],"prime": [200]})
df_2023
```

```python
df_2024 = pd.DataFrame({"nom": ["Bob"], "prime": [350]})
df_2024
```

```python
pd.concat([df_2023, df_2024])  #On remarque qu;apres concatenation les deu lignes on meme index.Donc il faut coriger
```

```python
pd.concat([df_2023, df_2024], ignore_index = True) # Pour corriger on affecte True a ignore_index ==>Ingnore les indices 
```

```python
## la concatenation peut se faire sous forme de ligne vertical ou horizontal , pour ce faire on utilise axis pour preciser

C= pd.concat([df_2023, df_2024], ignore_index = True, axis = 0) ## Concatenation  sur colone

L = pd.concat([df_2023, df_2024], ignore_index = True, axis = 1) ## Concatenation  sur une ligne
```

## En résumé:

    Il existe deux façons de fusionner deux data frames : 

        si les data frames ont la même structure, on peut faire une concaténation via la fonction concat : mettre les 2 data frames bout à bout ;

        sinon, on peut faire une jointure via la fonction/méthode << merge >> .

    Une jointure permet de joindre les informations de 2 data frames à partir d’une clé, une variable commune aux 2 data frames.

    Il existe 4 types de jointures : 

       - la jointure interne qui conserve les clés se trouvant dans le premier ET le second data frame ;

       - la jointure à droite (ou à gauche) qui conserve uniquement les clés se trouvant dans le data frame à droite (ou à gauche),
           et complète les informations manquantes par des valeurs manquantes, ou NaN ;

       - la jointure externe qui conserve toutes les clés se trouvant dans le premier OU le second data frame.

### Solution :

### (1) importation des deux fichier 

```python
clients_1 = pd.read_csv('https://raw.githubusercontent.com/OpenClassrooms-Student-Center/fr-4452741-decouvrez-les-librairies-python-pour-la-data-science/main/data/clients.csv')
clients_1.head()
```

```python
clients_2 = pd.read_csv('https://raw.githubusercontent.com/OpenClassrooms-Student-Center/fr-4452741-decouvrez-les-librairies-python-pour-la-data-science/main/data/clients_suite.csv')
clients_2.head()
```

<h3 align = "center">(1) -La première mission va être de rassembler les deux dataframes clients_1 et clients_2 en un gros dataframe clients qui contiendra l'ensemble des informations de notre clientèle !</h3>

```python
Dataset = pd.concat([clients_1 , clients_2], ignore_index = True) # Ici j'utilise concat pour concatener les deux dataframes car il ont une meme stricture
```

```python
Dataset.head()
```

```python
Dataset.shape[0] # Nombre totales de clients apres concatenation
```

<h3 align = "center"> (2) -A présent, vous allez fusionner ce fichier clients avec les profils clients créés précédemment. Ces derniers se trouve dans le dataframe profil_clients qui a été généré en case 2. Vous pouvez nommer le dataframe final data :</h3>

```python
profil_clients = pd.read_csv("profil_clients.csv") # importation du  dataset contenant le profil des clients
profil_clients.head()
```

```python
## Fusions des donnes dataset et leur profil

data = pd.merge(Dataset,profil_clients , on = "identifiant",how = "inner")
data.head()
```

<h3 align = "center"> (3) -Le service marketing de la banque nous a fourni un fichier dans lequel se trouve l'age de TOUS les clients de notre banque.</h3>

```python
clients_age = pd.read_csv('https://raw.githubusercontent.com/OpenClassrooms-Student-Center/fr-4452741-decouvrez-les-librairies-python-pour-la-data-science/main/data/client_age.csv')
clients_age.head() #import du fichier contenant les ages de nos clients
```

<p>Ajoutez l'information de l'age au dataframe data. Cependant, il semble que certains des clients ayant contracté un prêt n'apparaissent pas dans ce fichier. On souhaite cependant garder toutes les informations de notre dataframe data : choisissez donc avec précaution les arguments !</p>

```python
df = pd.merge(data, clients_age, on = "identifiant", how = "left") # Ici, l'enoncer nous 
#dit que il y'a certaain client qui n;ont pas communiquer leur ages  donc je vais garder tout les clients
```
