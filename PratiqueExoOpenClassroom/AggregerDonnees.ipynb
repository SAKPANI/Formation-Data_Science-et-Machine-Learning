```python
prets = pd.read_csv("Donnes_Banques.csv")
prets.head()
```

<h2 align = "center"> (1) Créer un data frame de profil client pour avoir par clienttoutes les informations qui le concernent, résumées en une ligne ;</h2> 

```python
profil_clients = prets.groupby('identifiant')[['remboursement','taux_endete','cout_total','benefices']].sum()
profil_clients.reset_index(inplace=True)
profil_clients.head()
```

### (2) calculer le nombre exact de personnes en état de situation bancaire risquée, à partir du taux d’endettement et de ces profils clients ;

```python
Taille = prets.loc[prets["taux_endete"]>35 , :].shape[0]
Taille
```

### (3) Calculer le bénéfice total dégagé par chacune de nos agences, par type de prêt ;

```python
prets.groupby("type")["benefices"].sum()
```

### (4) calculer les bénéfices moyens réalisés par chaque agence, pour chaque type de prêt, sous forme de tableau à double entrée ;

```python
prets.groupby(["ville", "type"])["benefices"].mean()
```

```python
Data = prets.pivot_table(index = "ville", columns ="type", values = "benefices", aggfunc = "mean")
Data
```

```python
Data.reset_index() #Retour au dataframe classique
```

### (5) me communiquer la ville qui semble la plus intéressante pour y développer les prêts immobiliers ?

```python
# Data = prets.groupby(["ville", "type"]).agg({"benefices" : "sum"})
# Data
```

```python
df = prets.pivot_table(index = "ville", columns ="type" , values = "benefices", aggfunc = "mean")
df
```

```python
df.sort_values("immobilier", ascending  = False)
```
