```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
# import du fichier client_assurance netoyee
df= pd.read_csv("fichier_netoyee.csv")
df
```

<h1 align="center" style="color:blue;">
 MATPLOTLIB
</h1>

```python
plt.plot(df["age"], df["prime"], label = "prime = f(age)")
plt.title("Prime en fonction de l'age")
plt.grid()
plt.xlabel("age")
plt.ylabel("prime")
plt.legend()
plt.show()
```

<h2 style ='color : blue'>A. Nuages de point (Relation entre deux variables) </h2>

<p>Pour analyser la relation entre deux variable on utilise les nuages de point.<br>La fonction qui permet de repenter les nuages de point avec la bibliotheque matplotlib est la fonction <span style = "color  : blue ">plt.scatter(var1, var2)</span></p>

```python
plt.scatter(df["age"], df["prime"], color ="blue")
plt.xlabel("age")
plt.ylabel("prime")
plt.title("Relation entre l'age et prime")
##plt.show()
```

```python
## d'APRES L'HISTOHRAMME PLUS L'AGE S'AUGMENTE , IL Y;A DIMUNITION DE PRIME
# - D'apres les nuages de points les jeunes obtiennes plus de primes que les vie
```

```python
##Corelation entre age et prime
```

```python
Coorelation  = df["age"].corr(df["prime"])
print("Correlation entre age et prime : ",float(Coorelation))
```

### Petite remarque sur les correlation :

#### - entre 0 et + - 0.2 ==> Correlation quasi inexistante
#### - Entre +-0.2 et +-0.4 ==> Correlation faible a modérée
#### - Entre +- 0.4 et +- 0.7 ==> Correlation moderee a forte 
#### - Au dela de +-0.7 ==> Correlation forte

<h2 style ='color : blue'>B. Histogramme (Distribution d'une variable) </h2>

```python
plt.hist(df["age"], bins =20, label = "age")  ##Decoupage en 20 histogramme (Par defaut python desssine 10)
plt.title("Distribution de l'ages des clients de l'assurance")
plt.legend()
plt.show()   ##La  distribution des ages 
```

<h2 style ='color : blue'>C. Diagramme en barres(comparer des categories) </h2>

```python
# Diagramme en barres(comparer des categories)
df["type_contrat"].value_counts().plot(kind="bar")
plt.title("Comparaison des categories d'une meme categorie( meme colonne)", color = "blue")
plt.show()
```

<h2 style ='color : blue'>D. Boite a moustache  </h2>

```python
# Boite a moustaches (detecter les outliers)
plt.boxplot(df["prime"].dropna())
plt.show()
```
