### A- LA FONCTION  " ***diag()*** "
#### Cette fonction nou permet de recuperer les diagonales d'une matrice 
#### on ecrit sous cette forme ***np.diag(matrice, k)***  avec :                            
#### - k>0 on prend les diagonales qui sont au dessus 
#### - Pour k<0 on prend les diagonales qui sont en dessous de la diagonale principale 
#### -Pour k=0 on retrouve la diagonale principale

```python
matrice = np.arange(1,17).reshape(4,4)
```

```python
matrice
```

```python
np.diag(matrice, k=0)
```

```python
np.diag(matrice,k= -1)
```

```python
np.diag(matrice,k= 3)
```

#### 1- Creation d'une matrice diagonale a partire d'un tableau a une dimension

```python
tab3=np.array([12,4,5,7,9])  ##CREATION DU TABLEAU A UNE SEUL DIMENSION
```

```python
tab3
```

```python
Matrice_Diagonale=np.diag(tab3)
```

```python
Matrice_Diagonale1=np.diag(tab3, k=-2)  ## nouvelle diagonale en dessous de la diagonale principale a un pas de 2
```

```python
Matrice_Diagonale1    
```
