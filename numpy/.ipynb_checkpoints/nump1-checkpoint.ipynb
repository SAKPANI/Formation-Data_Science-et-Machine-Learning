### 5-  LA FONCTION " diag() "
#### Cette fonction nou permet de recuperer les diagonales d'une matrice
#### on ecrit sous cette forme np.diag(matrice, k) avec :
#### - k>0 on prend les diagonales qui sont au dessus
#### - Pour k<0 on prend les diagonales qui sont en dessous de la diagonale principale
#### -Pour k=0 on retrouve la diagonale principale

```python
np.diag(A, k= 0)
```

```python
np.diag(A, k=-1)
```
