#### 1 - ) Calcule de  la moyenne de chaque matière.

```python
#Pour ce faire d'abord je vais fusionner les deux dataframes notes_obtenues, et cours 
Data_notes = pd.merge(cours, notes_obtenues, on = "course_id", how = "inner")
```

```python
Data_notes.head()
```

```python
Data_notes["matiere"].unique()  # Element de la colonne de matiere sans doublons
```

```python
# ha moi qui voulais utiliser la methode ci-dessus pour le calcul de moyenne de chaque matiere , tchieer, jamais de la vie , boucle for ma preférée

Dico_moyenne ={}
for i in Data_notes["matiere"].unique():
    moy = Data_notes.loc[Data_notes["matiere"] == i, "note"].mean()
    moyenne = round(float(moy), 2)
    Dico_moyenne[i] = moyenne
    
Dico_moyenne
```

```python
df_moyenne = pd.DataFrame(Dico_moyenne , index =["moyenne"])
```

#### 3-) Trouver la matière ayant la plus faible moyenne.

`|Pour ce faire je peut bien preciser la plus faible moyenne a partir du dataframe df_moyenne mais , bon je vais le determiner 
a partit d'un programme python en partant du dictionnaire Dico_moyenne`

```python
moy_min = list(Dico_moyenne.values())[0] # initialisation de la  premiere moyenne  du dictionnaire
matiere_min = ""

for matiere , moyenne in Dico_moyenne.items():
    if moyenne < moy_min:
        moy_min = moyenne
        matiere_min = matiere
        
print(f"Moyenne , {matiere_min} :{moy_min}")
```
