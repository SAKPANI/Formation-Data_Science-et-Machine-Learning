```python



# ##Maintenant on ne peut pas laisser les element null Comme cela on doit trouver un moyen de les remplacer :

#    - dans notre cas region a des NAN : je vais remplacer par (inconnues)  car  c'est un string
#    - prime contient des float je vais remplaceer les element manquant par la madiane  de la colonne
#    - Pour l'age je vais remplacer par la moyenne de tout les ages de la colonne 
```

```python
df["region"] = df["region"].fillna("inconue")   ##complete les case manquante de region par inconneue
```

```python
##Verification  avec .isnull.sum()
df.isnull().sum()
```

```python
#    On remarque que region est redevenu 0 donc pus de case sans element rexplorons le dataFrame entre [50 et 100] voir 

df.loc[50 :100]  ## On voit dans certains case c'est ecrit inconnue donc fil na a fonctionner 
```

```python
##Corection des case manquante de l'age .On remplace par la moyenne des ages

df["age"] = df["age"].fillna(df["age"].mean())
```

```python
df.isnull().sum()  ##On trouve 0 pour age  donc la methode a fonctionner il ne reste plus que pour prime 
```

```python
##Correction des NAN DE PRIME PAR LA MEDIAN

df["prime"] = df["prime"].fillna(df["prime"].median())
```
